```python
landing_outcomes = df['Outcome'].value_counts(); landing_outcomes
```

```python
bad_outcomes = set(landing_outcomes.keys()[[1,3,5,6,7]]); bad_outcomes
```

```python
df['Class'] = [0 if o in bad_outcomes else 1 for o in df['Outcome']]
print('Success rate:', df['Class'].mean())
df.to_csv('dataset_part_2.csv', index=False)
```
