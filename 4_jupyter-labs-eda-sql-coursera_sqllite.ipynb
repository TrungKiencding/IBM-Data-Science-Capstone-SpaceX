# SpaceX Falcon 9 – EDA with SQL (SQLite)

```python
import sqlite3, pandas as pd
con = sqlite3.connect('my_data1.db')
pd.read_csv('https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/labs/module_2/data/Spacex.csv').to_sql('SPACEXTBL', con, if_exists='replace', index=False)
q = lambda s: pd.read_sql(s, con)
```

Task 1 – Unique launch sites

```python
q('SELECT DISTINCT Launch_Site FROM SPACEXTBL')
```

Task 2 – 5 records where launch site begins with 'CCA'

```python
q("SELECT * FROM SPACEXTBL WHERE Launch_Site LIKE 'CCA%' LIMIT 5")
```

Task 3 – Total payload for NASA (CRS)

```python
q("SELECT SUM(PAYLOAD_MASS__KG_) AS Total FROM SPACEXTBL WHERE Customer='NASA (CRS)'")
```

Task 4 – Average payload for F9 v1.1

```python
q("SELECT AVG(PAYLOAD_MASS__KG_) AS Avg FROM SPACEXTBL WHERE Booster_Version='F9 v1.1'")
```

Task 5 – First successful ground pad landing

```python
q("SELECT MIN(Date) FROM SPACEXTBL WHERE Landing_Outcome='Success (ground pad)'")
```

Task 6 – Drone ship success, payload 4000–6000

```python
q("SELECT Booster_Version FROM SPACEXTBL WHERE Landing_Outcome='Success (drone ship)' AND PAYLOAD_MASS__KG_>4000 AND PAYLOAD_MASS__KG_<6000")
```

Task 7 – Mission outcomes

```python
q('SELECT Mission_Outcome, COUNT(*) AS Total FROM SPACEXTBL GROUP BY Mission_Outcome')
```

Task 8 – Boosters with max payload

```python
q('SELECT DISTINCT Booster_Version FROM SPACEXTBL WHERE PAYLOAD_MASS__KG_=(SELECT MAX(PAYLOAD_MASS__KG_) FROM SPACEXTBL)')
```

Task 9 – 2015 drone ship failures

```python
q("SELECT substr(Date,6,2) AS Month, Landing_Outcome, Booster_Version, Launch_Site FROM SPACEXTBL WHERE substr(Date,1,4)='2015' AND Landing_Outcome='Failure (drone ship)'")
```

Task 10 – Rank landing outcomes 2010-06-04 to 2017-03-20

```python
q("SELECT Landing_Outcome, COUNT(*) AS Cnt FROM SPACEXTBL WHERE Date BETWEEN '2010-06-04' AND '2017-03-20' GROUP BY Landing_Outcome ORDER BY Cnt DESC")
```
