# SpaceX Falcon 9 – EDA with Data Visualization

```python
import pandas as pd, seaborn as sns, matplotlib.pyplot as plt
df = pd.read_csv('https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_2.csv')
```

```python
sns.catplot(y='LaunchSite', x='FlightNumber', hue='Class', data=df, aspect=3); plt.show()
```

```python
sns.catplot(y='LaunchSite', x='PayloadMass', hue='Class', data=df, aspect=3); plt.show()
```

```python
df.groupby('Orbit')['Class'].mean().sort_values().plot(kind='bar'); plt.ylabel('Success rate'); plt.show()
```

```python
sns.catplot(y='Orbit', x='FlightNumber', hue='Class', data=df, aspect=3); plt.show()
```

```python
sns.catplot(y='Orbit', x='PayloadMass', hue='Class', data=df, aspect=3); plt.show()
```

```python
df['Year']=pd.to_datetime(df['Date']).dt.year; df.groupby('Year')['Class'].mean().plot(marker='o'); plt.ylabel('Success rate'); plt.show()
```

```python
features = df[['FlightNumber','PayloadMass','Orbit','LaunchSite','Flights','GridFins','Reused','Legs','LandingPad','Block','ReusedCount','Serial']]
features_one_hot = pd.get_dummies(features, columns=['Orbit','LaunchSite','LandingPad','Serial']).astype('float64')
features_one_hot.to_csv('dataset_part_3.csv', index=False); features_one_hot.shape
```
